## Imports

```python
import pygmt
import pandas as pd
```

## Mapping function

```python
# Function to create a map
def create_map(lons, lats, filename, region_vals, country):
    # Define data
    topo_data = pygmt.datasets.load_earth_relief(resolution='03s', region=region_vals)

    # Create a basemap
    fig = pygmt.Figure()

    # Make color palette
    pygmt.makecpt(cmap='topo', series='-8000/8000/1000', continuous=True)

    # Plot high resolution topography
    fig.grdimage(grid=topo_data, region=region_vals, projection='M4i', shading=True)

    # Plot coastlines and color the water
    water_color = '#4A90E2'
    fig.coast(region=region_vals, projection='M4i', shorelines=True, water=water_color, rivers=f"a/1p,{water_color}", lakes=water_color)

    # Plot topography contour lines
    fig.grdcontour(grid=topo_data, levels=4000, annotation="4000+f6p", pen="a0.15p")

    # Plot data points
    fig.plot(x=lons, y=lats, style='c0.07i', fill='red', pen='black')

    # Add title and axis labels
    with pygmt.config(FONT="10p", FIG_WIDTH="8i", FIG_HEIGHT="6i"):
        fig.basemap(frame=["afg"])

    # Plot colorbar
    fig.colorbar(frame='+lTopography')

    # Set inset region and country code based on input
    if country == "BO":  # Bolivia
        inset_region = [-72, -56, -24, -9]
        dcw_code = "BO"
    else: # Panama
        inset_region = [-84, -77, 7, 10]
        dcw_code = "PA"

    # Add inset map
    with fig.inset(position="jBR+o0.1c", box="+gwhite+p1p", region=inset_region, projection="M2c"):
        fig.coast(dcw=f"{dcw_code}+gyellowgreen+p0.2p", area_thresh=10000)
        rectangle = [[region_vals[0], region_vals[2], region_vals[1], region_vals[3]]]
        fig.plot(data=rectangle, style="r+s", pen=".5p,red")

    # Save the figure
    fig.savefig(filename)
```

## Map actual locations

```python
##### PANAMA MAP #####

# Get data
df = pd.read_csv('../data/data_panama/Panama_Plots_Metadata_all.csv', encoding='mac_roman')

# Collect all actual locations
all_lats_actual = df['Latitude']
all_lons_actual = df['Longitude']

# Create map for actual locations
create_map(all_lons_actual, all_lats_actual, 'figures/actual_locations_Panama.png', [-80.05, -79.3, 8.88, 9.39], country="PA")
```
