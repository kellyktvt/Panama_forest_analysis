```python
import h5py
import pandas as pd
```

## Feature x Embedding
For the sirius.mgf file and matrix, the embeddings would have to be averaged to give a single consensus score for each feature, since it includes repeat spectra from each sample run. 
For the gnps mgf, we already generated a concensus spectra.

```python
# with h5py.File('data/data_panama/PanamaProp_f100_20250105_sirius.hdf5', 'r') as f:
with h5py.File('data/data_panama/PanamaProp_f100_20250105_iimn_gnps.hdf5', 'r') as f:
    print(list(f.keys()))
```

```python
# Open HDF5 and extract selected datasets
# with h5py.File('data/data_panama/PanamaProp_f100_20250105_sirius.hdf5', 'r') as f:
with h5py.File('data/data_panama/PanamaProp_f100_20250105_iimn_gnps.hdf5', 'r') as f:
    feature_ids = f['FEATURE_ID'][:]
    # filenames = [x.decode('utf-8') for x in f['FILENAME'][:]]
    embs = f['DreaMS_embedding'][:]

# Create DataFrame
df_dreams = pd.DataFrame({
    'FEATURE_ID': feature_ids,
    # 'FILENAME': filenames,
    'embedding': list(embs)
})

# Count unique feature IDs
unique_feature_count = df_dreams['FEATURE_ID'].nunique()
print(f"Unique feature ID count: {unique_feature_count}")

df_dreams
```

## Feature x Species
Matrix where rows are individual compounds identified by FEATURE_ID.
- columns FEATURE_ID:molecularFormula provide chemical information about a specific FEATURE_ID - similar to how to the embeddings characterizes a FEATURE_ID
- columns acalma:virose is the TIC (abundance) of that feature in that sample

```python
# Read the mnemonic mastertable and assign it to a pandas dataframe
df_featureXspecies = pd.read_csv('data/data_panama/Panama2020method_mastertable_mnemonic_20250602.tsv', sep='\t')
df_featureXspecies
```

## Feature x Plot
Matrix where rows are individual compounds identified by FEATURE_ID.
- columns FEATURE_ID:molecularFormula provide chemical information about a specific FEATURE_ID - similar to how to the embeddings characterizes a FEATURE_ID
- columns BCI:metrop is the TIC abundance of each feature weighted by the abundance of each species in each plot

```python
# Read the plot mastertable and assign it to a pandas dataframe
df_featureXplot = pd.read_csv('data/data_panama/Panama2020method_mastertable_plot_20250602.tsv', sep='\t')
df_featureXplot
```

## Species x Plot
Matrix where rows = mnemonic_id, columns = individual plots. This give the abundance of each species in each plot where each species is identified by its mnemonic_id

```python
# Read the abundance matrix and assign it to a pandas dataframe
df_species_abundance = pd.read_csv('data/data_panama/Panama_Plots_Species_Abundance_Matrix.csv', sep=',')
df_species_abundance
```

## Plot Info
Gives the location, rainfall etc of each plot
- plot_name is the unique identifier and matches the columns in (2-Panama_Plots_Species_Abundance_Matrix)

```python
# Read the plot info and assign it to a pandas dataframe
df_plots = pd.read_csv('data/data_panama/Panama_Plots_Metadata_all.csv', sep=',', encoding='mac_roman')

# Remove the 'Unnamed: 0' column 
df_plots.drop(columns=['Unnamed: 0'], inplace=True)

# Change column name 'plot_name' to 'plot'
df_plots.rename(columns={'plot_name': 'plot'}, inplace=True)
df_plots
```

# Merge Data

### Feature x Embedding x Species

Merge <u>Feature x Embedding</u> with <u>Feature x Species</u>

```python
# Merge df_dreams with df_featureXspecies on FEATURE_ID
df_dreams_species = pd.merge(df_dreams, df_featureXspecies, on='FEATURE_ID', how='left')

# Identify species columns (acalma:virose)
species_columns = df_dreams_species.loc[:, 'acalma':'virose'].columns

# Identify all columns to keep as id_vars (all except species columns)
id_vars = [col for col in df_dreams_species.columns if col not in species_columns]

# Melt the DataFrame
df_dreams_species_melted = df_dreams_species.melt(
    id_vars=id_vars,
    value_vars=species_columns,
    var_name='species',
    value_name='feature_abundance'
)

# Filter out rows where abundance is zero
df_dreams_species_melted = df_dreams_species_melted[df_dreams_species_melted['feature_abundance'] > 0]
df_dreams_species_melted
```

```python
# Print columns with no values
empty_columns = df_dreams_species_melted.columns[df_dreams_species_melted.isnull().all()]
print("Columns with no values:", empty_columns.tolist())

# Remove empty columns
df_dreams_species_melted = df_dreams_species_melted.drop(columns=empty_columns)
df_dreams_species_melted
```

### Feature x Embedding x Plot
Merge Feature x Embedding with Feature x Plot

```python
# Merge df_dreams with df_featureXplot on FEATURE_ID
df_dreams_plot = pd.merge(df_dreams, df_featureXplot, on='FEATURE_ID', how='left')

# Identify plot columns (BCI:metrop)
plot_columns = df_dreams_plot.loc[:, 'BCI':'metrop'].columns

# Identify all columns to keep as id_vars (all except species columns)
id_vars = [col for col in df_dreams_plot.columns if col not in plot_columns]

# Melt the DataFrame
df_dreams_plot_melted = df_dreams_plot.melt(
    id_vars=id_vars,
    value_vars=plot_columns,
    var_name='plot',
    value_name='featureXspecies_abundance'
)

# Filter out rows where abundance is zero
df_dreams_plot_melted = df_dreams_plot_melted[df_dreams_plot_melted['featureXspecies_abundance'] > 0]
df_dreams_plot_melted
```

```python
# print columns with no values
empty_columns = df_dreams_plot_melted.columns[df_dreams_plot_melted.isnull().all()]
print(f"Columns with no values: {empty_columns.tolist()}")

# Remove empty columns
df_dreams_plot_melted = df_dreams_plot_melted.drop(columns=empty_columns)
df_dreams_plot_melted
```

### Feature x Embedding x Plot x Plot Info
Merge <u>Feature x Embedding x Plot</u> with <u>Plot Info</u>

```python
# Merge df_dreams_plot_melted with df_plots
df_dreams_plot_melted_info = pd.merge(df_dreams_plot_melted, df_plots, on='plot', how='left')
df_dreams_plot_melted_info 
```

# Save data

```python
# Convert the 'embedding' column to comma-separated strings
df_dreams_plot_melted_info['embedding'] = df_dreams_plot_melted_info['embedding'].apply(lambda x: ','.join(map(str, x)))

# Save df_dreams_plot_melted_info to a CSV file
df_dreams_plot_melted_info.to_csv('data/data_panama/Panama_gnps_dreams_plot_melted_info.csv', index=False)
```
